```python
import cv2
import time
from pathlib import Path
import matplotlib.pyplot as plt
from IPython.display import clear_output

# Camera IDs for left and right cameras
LEFT_CAMERA_ID = 1  
RIGHT_CAMERA_ID = 0

width = 1920
height = 1080

# Create output directory if it doesn't exist
output_dir = Path("logitech_brio_100_stereo2")
output_dir.mkdir(parents=True, exist_ok=True)

# Initialize cameras
left_cam = cv2.VideoCapture(LEFT_CAMERA_ID)
right_cam = cv2.VideoCapture(RIGHT_CAMERA_ID)
left_cam.set(cv2.CAP_PROP_FRAME_WIDTH, width)
left_cam.set(cv2.CAP_PROP_FRAME_HEIGHT, height)
right_cam.set(cv2.CAP_PROP_FRAME_WIDTH, width)
right_cam.set(cv2.CAP_PROP_FRAME_HEIGHT, height)

frame_count = 0
timer = 0
WAIT = 0.05
EVERY = 2

try:
    while True:
        # Capture frames from both cameras
        ret_left, left_frame = left_cam.read()
        ret_right, right_frame = right_cam.read()
        
        if not ret_left or not ret_right:
            print("Failed to capture frames")
            break

        # Render frames
        plt.subplot(1, 2, 1)
        plt.axis('off') 
        plt.title('Left Camera')
        plt.imshow(cv2.cvtColor(left_frame, cv2.COLOR_BGR2RGB))

        plt.subplot(1, 2, 2)
        plt.axis('off') 
        plt.title('Right Camera')
        plt.imshow(cv2.cvtColor(right_frame, cv2.COLOR_BGR2RGB))

        clear_output(wait=True)
        plt.show()

        if timer >= EVERY:
            # Save frames
            timestamp = int(time.time() * 1000)  # Get current timestamp in milliseconds
            left_filename = output_dir / "left" / f"{timestamp}.png"
            right_filename = output_dir / "right"/ f"{timestamp}.png"
            
            cv2.imwrite(str(left_filename), left_frame)
            cv2.imwrite(str(right_filename), right_frame)
            
            print(f"Saved frame pair {frame_count}")
            frame_count += 1
            timer = 0
        
        # Wait 5 seconds before next capture
        time.sleep(WAIT)
        timer += WAIT

finally:
    # Release cameras and close windows
    left_cam.release()
    right_cam.release()
```
