```python
import cv2
import time
from pathlib import Path
import matplotlib.pyplot as plt
from IPython.display import clear_output
from picamera2 import Picamera2

# Camera IDs for left and right cameras
LEFT_CAMERA_ID = 1  
RIGHT_CAMERA_ID = 0

width = 1920
height = 1080

# Create output directory if it doesn't exist
output_dir = Path("IMX219_stereo")
output_dir.mkdir(parents=True, exist_ok=True)
(output_dir / "left").mkdir(parents=True, exist_ok=True)
(output_dir / "right").mkdir(parents=True, exist_ok=True)

# Initialize cameras
picam_left = Picamera2(LEFT_CAMERA_ID)
picam_right = Picamera2(RIGHT_CAMERA_ID)
for cam in [picam_left, picam_right]:
    cam.preview_configuration.main.size = (width, height)
    cam.preview_configuration.main.format = "YUV420"
    cam.configure("preview")
    cam.start()


frame_count = 0
timer = time.time()
EVERY = 10

try:
    while True:
        # Capture frames from both cameras
        left_frame = picam_left.capture_array()
        right_frame = picam_right.capture_array()

        # Convert
        left_frame = cv2.cvtColor(left_frame, cv2.COLOR_YUV2RGB_I420)
        right_frame = cv2.cvtColor(right_frame, cv2.COLOR_YUV2RGB_I420)

        # Render frames
        plt.figure(figsize=(15,8))
        
        plt.subplot(1, 2, 1)
        plt.axis('off') 
        plt.title('Left Camera')
        plt.imshow(left_frame)

        plt.subplot(1, 2, 2)
        plt.axis('off') 
        plt.title('Right Camera')
        plt.imshow(right_frame)

        clear_output(wait=True)
        plt.show()

        if time.time()-timer >= EVERY:
            # Save frames
            timestamp = int(time.time() * 1000)  # Get current timestamp in milliseconds
            left_filename = output_dir / "left" / f"{timestamp}.png"
            right_filename = output_dir / "right"/ f"{timestamp}.png"
            
            cv2.imwrite(str(left_filename), cv2.cvtColor(left_frame, cv2.COLOR_RGB2BGR))
            cv2.imwrite(str(right_filename), cv2.cvtColor(right_frame, cv2.COLOR_RGB2BGR))
            
            print(f"Saved frame pair {frame_count}")
            frame_count += 1
            time.sleep(0.5)
            timer = time.time()

finally:
    # Release cameras and close windows
    picam_left.close()
    picam_right.close()
```
